```python
def smart_parking_system(capacity, vehicle_logs):
    current_parked = 0

    for log in vehicle_logs:
        if log == "IN":
            current_parked += 1
        elif log == "OUT" and current_parked > 0:
            current_parked -= 1

    if current_parked > capacity:
        status = "Alert"
    else:
        status = "Available"

    print("Currently Parked Vehicles:", current_parked)
    print("Parking Status:", status)


# Sample Input
smart_parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5  # 5 minutes per km → 8*5 = 40

    if traffic == "High":
        base_time += 10
    elif traffic == "Medium":
        base_time += 5

    if weather == "Rainy":
        base_time += 5

    print("Estimated Delivery Time:", base_time, "minutes")


# Sample Input
estimate_delivery_time(8, "High", "Rainy")
```

```python
def seat_occupancy_analyzer(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    # Determine show status
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    print("Occupancy:", round(occupancy), "%")
    print("Show Status:", status)


# Sample Input
seat_occupancy_analyzer(200, [1]*150)
```

```python
def classify_server_load(cpu_readings):
    total = 0

    # Calculate total CPU usage
    for reading in cpu_readings:
        total += reading

    average = total / len(cpu_readings)

    # Classification
    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", round(average), "%")
    print("Server Status:", status)


# Sample Input
classify_server_load([45, 60, 70, 85, 90])
```

```python
def resource_usage_monitor(resource_usage):
    overused = []
    threshold = 8  # Overuse threshold in hours

    # Check each resource usage
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused.append(resource)

    # Energy alert condition
    if overused:
        alert = "Yes"
    else:
        alert = "No"

    print("Overused Resources:", ", ".join(overused) if overused else "None")
    print("Energy Alert:", alert)


# Sample Input
resource_usage_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

```python
def registration_controller(capacity, registrations):
    
    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
registration_controller(100, 105)
```
